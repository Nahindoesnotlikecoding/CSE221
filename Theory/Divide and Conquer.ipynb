```python
# Merge sort
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr)//2
        left_arr = arr[:mid]
        right_arr = arr[mid:]

        # Recursion part
        merge_sort(left_arr)
        merge_sort(right_arr)

        # Merge part
        i = 0
        j = 0
        k = 0

        while i<len(left_arr) and j<len(right_arr):
            if left_arr[i] < right_arr[j]:
                arr[k] = left_arr[i]
                i+=1
            else:
                arr[k] = right_arr[j]
                j += 1

            k += 1

        while i<len(left_arr):
            arr[k] = left_arr[i]
            i+=1
            k+=1

        while j<len(right_arr):
            arr[k] = right_arr[j]
            j+=1
            k+=1
    return arr


arr = [3, 5, 2, 1, 7, 8, 4, 9, 6]
print(f"Unsorted Array: {arr}")
sorted_arr = merge_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
def partition(array, lo, hi):
  pivot = arr[lo]

  i = lo+1

  for j in range(lo+1, hi+1):
    if arr[j]<pivot:
      arr[j], arr[i] = arr[i], arr[j]
      i += 1

  arr[lo], arr[i-1] = arr[i-1], arr[lo]

  return i-1




def quick_sort(arr, lo, hi):
  if lo<hi:
    pivot_idx = partition(arr, lo, hi)
    quick_sort(arr, lo, pivot_idx-1)
    quick_sort(arr, pivot_idx+1, hi)

  return arr


arr = [3, 5, 2, 1, 7, 8, 4, 9, 6]
print(f"Unsorted Array: {arr}")
sorted_arr = quick_sort(arr, 0, len(arr)-1)
print(f"Sorted Array: {sorted_arr}")
```

```python
def cross_sum_subarray(arr, left, right):
  mid = (left+right)//2
  left_sum = 0
  sum = 0
  for i in range(mid, left-1, -1):
    sum += arr[i]
    left_sum = max(sum, left_sum)

  sum = 0
  right_sum = 0
  for i in range(mid+1, right+1):
    sum += arr[i]
    right_sum = max(sum, right_sum)

  return left_sum + right_sum


def max_sum_subarray(arr, left, right):
  if left==right:
    return arr[left]

  mid = (left+right)//2

  lss = max_sum_subarray(arr, left, mid)
  rss = max_sum_subarray(arr, mid+1, right)
  css = cross_sum_subarray(arr, left, right)

  return max(lss, rss, css)

arr = [4, -18, -8, -2, 16, -14, 12, -1, 3, 15]
res = max_sum_subarray(arr, 0, len(arr)-1)
print(res)
```
