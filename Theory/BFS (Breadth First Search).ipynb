```python
# This version of Queue is more efficient
from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        else:
            return self.items.popleft()

    def is_empty(self):
        if len(self.items)==0:
          return True
        return False
```

```python
class Vertex:
    def __init__(self, key):
        self.key = key
        self.color = "white"
        self.distance = float('inf')
        self.parent = None
```

```python
class Graph:
    def __init__(self):
        # Map keys to Vertex objects.
        self.vertices = {}
        # Adjacency list: mapping vertex key -> list of adjacent Vertex objects.
        self.adj_list = {}

    def add_vertex(self, key):
        v = Vertex(key)
        self.vertices[key] = v
        self.adj_list[key] = []

    def add_edge(self, from_key, to_key, directed=True):
        self.adj_list[from_key].append(self.vertices[to_key])
        if directed==False:
          self.adj_list[to_key].append(self.vertices[from_key])
```

```python
def bfs(graph, start_key):
    # Initialize all vertices.
    for key in graph.vertices:
        v = graph.vertices[key]
        v.color = "white"
        v.distance = float('inf')
        v.parent = None

    # Set up the starting vertex.
    start_vertex = graph.vertices[start_key]
    start_vertex.color = "gray"
    start_vertex.distance = 0
    start_vertex.parent = None

    # Initialize a custom FIFO queue and enqueue the start vertex.
    Q = Queue()
    Q.enqueue(start_vertex)

    # BFS main loop: process vertices until the queue is empty.
    while not Q.is_empty():
        u = Q.dequeue()
        # For each adjacent vertex of u.
        for v in graph.adj_list[u.key]:
            if v.color == "white":
                v.color = "gray"
                v.distance = u.distance + 1
                v.parent = u
                Q.enqueue(v)
        u.color = "black"
```

```python
g = Graph()

# Adding vertices (explicit addition is optional, as add_edge creates missing vertices).
for key in ['A', 'B', 'C', 'D', 'E', 'F']:
    g.add_vertex(key)

# Add directed edges.
g.add_edge('A', 'B')
g.add_edge('A', 'C')
g.add_edge('B', 'D')
g.add_edge('B', 'E')
g.add_edge('C', 'F')
g.add_edge('E', 'F')

# Run BFS starting from vertex 'A'.
bfs(g, 'A')
```

## To find the shortest path using BFS in an unweighted graph

```python
def shortest_path(graph, start_key, goal_key):
    # Run BFS to set the parent pointers.
    bfs(graph, start_key)

    # If the goal vertex has an infinite distance, it's unreachable.
    if graph.vertices[goal_key].distance == float('inf'):
        return None

    # Reconstruct the path by backtracking from the goal.
    path = [goal_key]
    current = graph.vertices[goal_key]
    while current.parent!=None:
        path.append(current.parent.key)
        current = current.parent

    for i in range(len(path)-1, -1, -1):
      if i==0:
        print(path[i])
      else:
        print(path[i], end="-->")
```

## Bi-partite Graph
* Bi-partite graph also means the graph has odd cycle

```python
class Graph:
    def __init__(self):
        # Map keys to Vertex objects.
        self.vertices = {}
        # Adjacency list: mapping vertex key -> list of adjacent Vertex objects.
        self.adj_list = {}

    def add_vertex(self, key):
        v = Vertex(key)
        self.vertices[key] = v
        self.adj_list[key] = []

    def add_edge(self, from_key, to_key, directed=True):
        self.adj_list[from_key].append(self.vertices[to_key])
        if directed==False:
          self.adj_list[to_key].append(self.vertices[from_key])
```

```python
def Bipartite(graph, start_key):
  for v in graph.vertices:
    v = graph.vertices[v]
    v.color = None

  start_vertex = graph.vertices[start_key]
  start_vertex.color = 0
  Q = Queue()
  Q.enqueue(start_vertex)

  while Q.is_empty() != True:
    u = Q.dequeue()
    for v in graph.adj_list[u.key]:
      if v.color == None:
        v.color = 1-u.color
        Q.enqueue(v)
      elif u.color == v.color:
        return "Not a bipartite Graph"

  return "Bipartite Graph"
```

```python
g = Graph()
for key in ["A", "B", "C", "D"]:
  g.add_vertex(key)

g.add_edge("A", "B")
g.add_edge("A", "C")

g.add_edge("B", "A")
g.add_edge("B", "D")

g.add_edge("D", "B")
g.add_edge("D", "C")

g.add_edge("C", "A")
g.add_edge("C", "D")

# g.add_edge("A", "D")

res = Bipartite(g, "A")
print(res)
```
