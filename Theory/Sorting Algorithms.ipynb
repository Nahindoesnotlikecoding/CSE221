```python
# Bubble sort
def bubble_sort(arr):
  for i in range(len(arr)):
    for j in range(len(arr)-1-i):
      if arr[j] > arr[j+1]:
        temp = arr[j]
        arr[j] = arr[j+1]
        arr[j+1] = temp

  return arr


arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = bubble_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# Selection sort
def selection_sort(arr):
  for i in range(0, len(arr)):
    min_index = i
    for j in range(i+1, len(arr)):
      if arr[min_index] > arr[j]:
        min_index = j

    temp = arr[i]
    arr[i] = arr[min_index]
    arr[min_index] = temp

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = selection_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# insertion sort
def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i-1
    while j >= 0 and key < arr[j]:
      arr[j+1] = arr[j]
      j -= 1
    arr[j+1] = key

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = insertion_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
class MaxHeap:
  def __init__(self, cap):
    self.arr = np.zeros(cap, dtype="int")
    self.size = 0
    self.cap = cap

  def insert(self, value):
    if self.size < self.cap:
      self.size += 1
      self.arr[self.size-1] = value
      self.swim(self.size-1)
    else:
      return

  def swim(self, index):
    if index>0:
      parent = (index-1)//2
      if self.arr[index] > self.arr[parent]:
        self.arr[index], self.arr[parent] = self.arr[parent], self.arr[index]
        self.swim(parent)
      else:
        return

  def extractMax(self):
    temp = self.arr[0]
    self.arr[0], self.arr[self.size-1] = self.arr[self.size-1], self.arr[0]
    self.size-=1
    self.arr[self.size] = 0
    self.sink(0)
    return temp

  def sink(self, index):
    if index<self.size:
      left = (2*index)+1
      right = (2*index)+2
      if left < self.size:
        if right < self.size:
          big = 0
          if self.arr[left] > self.arr[right]:
            big = left
          else:
            big = right
          if self.arr[big] > self.arr[index]:
            self.arr[big], self.arr[index] = self.arr[index], self.arr[big]
            self.sink(big)
        else:
          if self.arr[left] > self.arr[index]:
            self.arr[left], self.arr[index] = self.arr[index], self.arr[left]
            self.sink(left)

    return
```

```python
# Heap sort
def heap_sort(MaxHeap, arr):
  sort_heap = MaxHeap(len(arr))
  for  i in range(len(arr)):
    sort_heap.insert(arr[i])
  for i in range(len(arr)-1, -1, -1):
    extracted = sort_heap.extractMax()
    arr[i] = extracted

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = heap_sort(MaxHeap, arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# count sort (stable algorithm)
def count_sort(arr):
  #step-1
  max = arr[0]
  for i in range(len(arr)):
    if arr[i] > max:
      max = arr[i]

  # step-2
  new_arr = np.zeros(max+1, dtype="int")

  # step-3
  for i in range(len(arr)):
    new_arr[arr[i]] += 1

  # step-4
  for i in range(1, len(new_arr)):
    new_arr[i] = new_arr[i] + new_arr[i-1]

  # step-5
  out_arr = np.zeros(len(arr), dtype="int")

  # step-6 (stable algorithm as the indexing in the loop is starting from the last position)
  for i in range(len(arr)-1, -1, -1):
    num = arr[i]
    new_arr[num] -= 1
    out_arr[new_arr[num]] = num

  return out_arr




# arr = np.array([2, 1, 2, 3, 4, 6, 5, 3, 2, 3])
arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = count_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# count sort (unstable algorithm)
def count_sort(arr):
  #step-1
  max = arr[0]
  for i in range(len(arr)):
    if arr[i] > max:
      max = arr[i]

  # step-2
  new_arr = np.zeros(max+1, dtype="int")

  # step-3
  for i in range(len(arr)):
    new_arr[arr[i]] += 1

  # step-4
  for i in range(1, len(new_arr)):
    new_arr[i] = new_arr[i] + new_arr[i-1]

  # step-5
  out_arr = np.zeros(len(arr), dtype="int")

  # step-6 (unstable algorithm as the indexing in the loop is starting from the first position)
  for i in range(0, len(arr)):
    num = arr[i]
    new_arr[num] -= 1
    out_arr[new_arr[num]] = num

  return out_arr




# arr = np.array([2, 1, 2, 3, 4, 6, 5, 3, 2, 3])
arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = count_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# Merge sort
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr)//2
        left_arr = arr[:mid]
        right_arr = arr[mid:]

        # Recursion part
        merge_sort(left_arr)
        merge_sort(right_arr)

        # Merge part
        i = 0
        j = 0
        k = 0

        while i<len(left_arr) and j<len(right_arr):
            if left_arr[i] < right_arr[j]:
                arr[k] = left_arr[i]
                i+=1
            else:
                arr[k] = right_arr[j]
                j += 1

            k += 1

        while i<len(left_arr):
            arr[k] = left_arr[i]
            i+=1
            k+=1

        while j<len(right_arr):
            arr[k] = right_arr[j]
            j+=1
            k+=1
    return arr


arr = [3, 5, 2, 1, 7, 8, 4, 9, 6]
print(f"Unsorted Array: {arr}")
sorted_arr = merge_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
def partition(array, lo, hi):
  pivot = arr[lo]

  i = lo+1

  for j in range(lo+1, hi+1):
    if arr[j]<pivot:
      arr[j], arr[i] = arr[i], arr[j]
      i += 1

  arr[lo], arr[i-1] = arr[i-1], arr[lo]

  return i-1




def quick_sort(arr, lo, hi):
  if lo<hi:
    pivot_idx = partition(arr, lo, hi)
    quick_sort(arr, lo, pivot_idx-1)
    quick_sort(arr, pivot_idx+1, hi)

  return arr


arr = [3, 5, 2, 1, 7, 8, 4, 9, 6]
print(f"Unsorted Array: {arr}")
sorted_arr = quick_sort(arr, 0, len(arr)-1)
print(f"Sorted Array: {sorted_arr}")
```
