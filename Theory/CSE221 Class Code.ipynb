```python
# Bubble sort
def bubble_sort(arr):
  for i in range(0, len(arr)):
    for j in range(1, len(arr)-i):
      if arr[j] < arr[j-1]:
        temp = arr[j]
        arr[j] = arr[j-1]
        arr[j-1] = temp

  return arr


arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = bubble_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# Selection sort
def selection_sort(arr):
  for i in range(0, len(arr)):
    min_index = i
    for j in range(i+1, len(arr)):
      if arr[min_index] > arr[j]:
        min_index = j

    temp = arr[i]
    arr[i] = arr[min_index]
    arr[min_index] = temp

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = selection_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# insertion sort
def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i-1
    while j >= 0 and key < arr[j]:
      arr[j+1] = arr[j]
      j -= 1
    arr[j+1] = key

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = insertion_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
class MaxHeap:
  def __init__(self, cap):
    self.arr = np.zeros(cap, dtype="int")
    self.size = 0
    self.cap = cap

  def insert(self, value):
    if self.size < self.cap:
      self.size += 1
      self.arr[self.size-1] = value
      self.swim(self.size-1)
    else:
      return

  def swim(self, index):
    if index>0:
      parent = (index-1)//2
      if self.arr[index] > self.arr[parent]:
        self.arr[index], self.arr[parent] = self.arr[parent], self.arr[index]
        self.swim(parent)
      else:
        return

  def extractMax(self):
    temp = self.arr[0]
    self.arr[0], self.arr[self.size-1] = self.arr[self.size-1], self.arr[0]
    self.size-=1
    self.arr[self.size] = 0
    self.sink(0)
    return temp

  def sink(self, index):
    if index<self.size:
      left = (2*index)+1
      right = (2*index)+2
      if left < self.size:
        if right < self.size:
          big = 0
          if self.arr[left] > self.arr[right]:
            big = left
          else:
            big = right
          if self.arr[big] > self.arr[index]:
            self.arr[big], self.arr[index] = self.arr[index], self.arr[big]
            self.sink(big)
        else:
          if self.arr[left] > self.arr[index]:
            self.arr[left], self.arr[index] = self.arr[index], self.arr[left]
            self.sink(left)

    return
```

```python
# Heap sort
def heap_sort(MaxHeap, arr):
  sort_heap = MaxHeap(len(arr))
  for  i in range(len(arr)):
    sort_heap.insert(arr[i])
  for i in range(len(arr)-1, -1, -1):
    extracted = sort_heap.extractMax()
    arr[i] = extracted

  return arr

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = heap_sort(MaxHeap, arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# Linear Search
def linear_search(arr, num):
  for i in range(len(arr)):
    if arr[i] == num:
      return f"Number found at index {i}"
  return

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = linear_search(arr, 9)
print(search)
```

```python
# Binary Search
def binary_search(arr, num):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1

  while lo <= hi:
    mid = (lo+hi)//2
    if arr[mid] == num:
      return f"Found at index {mid}"
    elif arr[mid] > num:
      hi = mid-1
    elif arr[mid] < num:
      lo = mid+1

  return "Not found"

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = binary_search(arr, 9)
print(search)
```

```python
# Ternary Search
def ternary_search(arr, num):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1

  while lo <= hi:
    mid1 = lo + ((hi - lo) // 3)
    mid2 = hi - ((hi - lo) // 3)
    if arr[mid1] > num:
      hi = mid1-1
    elif arr[mid1]<num<arr[mid2]:
      lo = mid1+1
      hi = mid2-1
    elif arr[mid2]<num:
      lo = mid2+1
    elif arr[mid1] == num:
      return f"Value found at index {mid1}"
    elif arr[mid2] == num:
      return f"Value found at index {mid2}"

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = ternary_search(arr, 9)
print(search)
```

```python
def first_last_occurence(arr, value):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1
  first = None
  last = None

  while lo <= hi:
    # Searching for the first element
    mid = (lo+hi)//2
    if arr[mid]==value:
      first = mid
      hi = mid-1
    elif arr[mid] > value:
      hi = mid-1
    elif arr[mid] < value:
      lo = mid+1

  lo = 0
  hi = len(arr)-1
  while lo <= hi:
    # Searching for the last element
    mid = (lo+hi)//2
    if arr[mid] == value:
      last = mid
      lo = mid+1
    elif arr[mid] > value:
      hi = mid-1
    elif arr[mid] < value:
      lo = mid+1


  print("First occurrence:", first)
  print("Last occurrence:", last)

arr = np.array([1, 2, 2, 3, 3, 3, 3, 4, 5, 7, 3, 6, 6, 7])
first_last_occurence(arr, 3)
```

```python
# count sort (stable algorithm)
def count_sort(arr):
  #step-1
  max = arr[0]
  for i in range(len(arr)):
    if arr[i] > max:
      max = arr[i]

  # step-2
  new_arr = np.zeros(max+1, dtype="int")

  # step-3
  for i in range(len(arr)):
    new_arr[arr[i]] += 1

  # step-4
  for i in range(1, len(new_arr)):
    new_arr[i] = new_arr[i] + new_arr[i-1]

  # step-5
  out_arr = np.zeros(len(arr), dtype="int")

  # step-6 (stable algorithm as the indexing in the loop is starting from the last position)
  for i in range(len(arr)-1, -1, -1):
    num = arr[i]
    new_arr[num] -= 1
    out_arr[new_arr[num]] = num

  return out_arr




# arr = np.array([2, 1, 2, 3, 4, 6, 5, 3, 2, 3])
arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = count_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
# count sort (unstable algorithm)
def count_sort(arr):
  #step-1
  max = arr[0]
  for i in range(len(arr)):
    if arr[i] > max:
      max = arr[i]

  # step-2
  new_arr = np.zeros(max+1, dtype="int")

  # step-3
  for i in range(len(arr)):
    new_arr[arr[i]] += 1

  # step-4
  for i in range(1, len(new_arr)):
    new_arr[i] = new_arr[i] + new_arr[i-1]

  # step-5
  out_arr = np.zeros(len(arr), dtype="int")

  # step-6 (unstable algorithm as the indexing in the loop is starting from the first position)
  for i in range(0, len(arr)):
    num = arr[i]
    new_arr[num] -= 1
    out_arr[new_arr[num]] = num

  return out_arr




# arr = np.array([2, 1, 2, 3, 4, 6, 5, 3, 2, 3])
arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
print(f"Unsorted Array: {arr}")
sorted_arr = count_sort(arr)
print(f"Sorted Array: {sorted_arr}")
```

```python
def find_transition_index(arr):
    left = 0
    right = len(arr) - 1
    while left < right:
        mid = (left + right) // 2
        if arr[mid] > arr[right]:
            left = mid + 1
        else:
            right = mid
    return left


arr = np.array([9, 12, 15, 17, 18, 1, 2, 4, 5, 7, 8])
i = find_transition_index(arr)
print("Transition index i =", i)
```
