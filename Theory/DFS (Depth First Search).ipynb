```python
class Vertex:
    def __init__(self, key):
        self.key = key
        self.color = "white"
        self.start_time = 0
        self.finish_time = 0
        self.parent = None
```

```python
class Graph:
    def __init__(self):
        # Map keys to Vertex objects.
        self.vertices = {}
        # Adjacency list: mapping vertex key -> list of adjacent Vertex objects.
        self.adj_list = {}

    def add_vertex(self, key):
        v = Vertex(key)
        self.vertices[key] = v
        self.adj_list[key] = []

    def add_edge(self, from_key, to_key, directed=True):
        self.adj_list[from_key].append(self.vertices[to_key])
        if directed==False:
          self.adj_list[to_key].append(self.vertices[from_key])
```

```python
def DFS_visit(graph, u):
  global time
  u.color = "gray"
  time += 1
  u.start_time = time
  for v in graph.adj_list[u.key]:
    if v.color == "white":
      DFS_visit(graph, v)

  u.color = "black"
  time += 1
  u.finish_time = time

def DFS(graph):
  for key in graph.vertices:
    u = graph.vertices[key]
    u.color = "white"

  global time
  time = 0

  for key in graph.vertices:
    u = graph.vertices[key]
    if u.color == "white":
      DFS_visit(graph, u)
```

## Cycle Detection

```python
def DFS_visit(graph, u):
  global time, count
  u.color = "gray"
  time += 1
  u.start_time = time
  for v in graph.adj_list[u.key]:
    if v.color == "white":
      DFS_visit(graph, v)
    elif v.color == "gray":
      count += 1
      print(f"{count} Cycle Detected")
  u.color = "black"
  time += 1
  u.finish_time = time

def cycle_detection(graph):
  global time, count

  for key in graph.vertices:
    u = graph.vertices[key]
    u.color = "white"
  time = 0
  count = 0
  for key in graph.vertices:
    u = graph.vertices[key]
    if u.color == "white":
      DFS_visit(graph, u)
```

## Topological Sort

```python
def DFS_visit(graph, stack, u):
  u.color = "gray"
  for v in graph.adj_list[u.key]:
    if v.color == "white":
      x = DFS_visit(graph, stack, v)
      if x == False:
        return x
    elif v.color == "gray":
      return False

  u.color = "black"
  stack.append(u.key)
  return True

def topological_sort(graph):
  stack = []
  for key in graph.vertices:
    u = graph.vertices[key]
    u.color = "white"

  for key in graph.vertices:
    u = graph.vertices[key]
    if u.color == "white":
      x = DFS_visit(graph, stack, u)
      if x==False:
        print("Cycle Detected; Not a DAG")
        return


  while stack:
    if len(stack)==1:
      print(stack.pop())
    else:
      print(stack.pop(), end="-->")
```

```python
g = Graph()
for key in ["A", "B", "C", "D", "E", "F", "G"]:
  g.add_vertex(key)

g.add_edge("A", "B")
g.add_edge("A", "C")

g.add_edge("B", "C")
g.add_edge("B", "D")

# g.add_edge("C", "D")

# g.add_edge("D", "A")

g.add_edge("D", "E")

g.add_edge("E", "G")

g.add_edge("F", "D")
g.add_edge("F", "G")

topological_sort(g)
```

## Strongly Connected Components (Kosaraju's Algortihm)

```python
def dfs_visit(graph, stack, u):
  u.color = "gray"
  for v in graph.adj_list[u.key]:
    if v.color == "white":
      dfs_visit(graph, stack, v)

  u.color = "black"
  stack.append(u.key)

def graph_transpose(graph):
  transposed_graph = Graph()
  for key in graph.vertices:
      transposed_graph.add_vertex(key)

  for key in graph.adj_list:
    edge_list = graph.adj_list[key]
    for u in edge_list:
      transposed_graph.add_edge(u.key, key)

  return transposed_graph

def scc_dfs_visit(transposed_graph, scc_stack, u):
  u.color = "gray"
  scc_stack.append(u.key)
  for v in transposed_graph.adj_list[u.key]:
    if v.color == "white":
      scc_dfs_visit(transposed_graph, scc_stack, v)

  u.color = "black"

def strongly_connected_components(graph):
  for key in graph.vertices:
    u = graph.vertices[key]
    u.color = "white"
  stack = []
  for key in graph.vertices:
    u = graph.vertices[key]
    if u.color == "white":
      dfs_visit(graph, stack, u)

  transposed_graph = graph_transpose(graph)
  for key in transposed_graph.vertices:
    u = transposed_graph.vertices[key]
    u.color = "white"

  scc_components = []
  while stack:
    key = stack.pop()
    u = transposed_graph.vertices[key]
    if u.color == "white":
      scc_stack = []
      scc_dfs_visit(transposed_graph, scc_stack, u)
      scc_components.append(scc_stack)

  return scc_components
```

```python
g = Graph()
for key in ["A", "B", "C", "D", "E", "F", "G", "H"]:
  g.add_vertex(key)

g.add_edge("A", "B")
g.add_edge("B", "D")
g.add_edge("C", "A")
g.add_edge("D", "C")

g.add_edge("D", "E")

g.add_edge("E", "F")
g.add_edge("E", "H")

g.add_edge("F", "G")

g.add_edge("G", "E")

scc_components = strongly_connected_components(g)
print(scc_components)
```
