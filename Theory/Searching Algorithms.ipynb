```python
import numpy as np
```

```python
# Linear Search
def linear_search(arr, num):
  for i in range(len(arr)):
    if arr[i] == num:
      return f"Number found at index {i}"
  return

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = linear_search(arr, 9)
print(search)
```

```python
# Binary Search
def selection_sort(arr):
  for i in range(0, len(arr)):
    min_index = i
    for j in range(i+1, len(arr)):
      if arr[min_index] > arr[j]:
        min_index = j

    temp = arr[i]
    arr[i] = arr[min_index]
    arr[min_index] = temp

  return arr

def binary_search(arr, num):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1

  while lo <= hi:
    mid = (lo+hi)//2
    if arr[mid] == num:
      return f"Value found at index {mid}"
    elif arr[mid] > num:
      hi = mid-1
    elif arr[mid] < num:
      lo = mid+1

  return "Not found"

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = binary_search(arr, 9)
print(search)
```

```python
# Ternary Search
def ternary_search(arr, num):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1

  while lo <= hi:
    mid1 = lo + ((hi - lo) // 3)
    mid2 = hi - ((hi - lo) // 3)
    if arr[mid1] > num:
      hi = mid1-1
    elif arr[mid1]<num<arr[mid2]:
      lo = mid1+1
      hi = mid2-1
    elif arr[mid2]<num:
      lo = mid2+1
    elif arr[mid1] == num:
      return f"Value found at index {mid1}"
    elif arr[mid2] == num:
      return f"Value found at index {mid2}"

arr = np.array([3, 5, 2, 1, 7, 8, 4, 9, 6])
search = ternary_search(arr, 9)
print(search)
```

```python
def first_last_occurence(arr, value):
  arr = selection_sort(arr)
  lo = 0
  hi = len(arr)-1
  first = None
  last = None

  while lo <= hi:
    # Searching for the first element
    mid = (lo+hi)//2
    if arr[mid]==value:
      first = mid
      hi = mid-1
    elif arr[mid] > value:
      hi = mid-1
    elif arr[mid] < value:
      lo = mid+1

  lo = 0
  hi = len(arr)-1
  while lo <= hi:
    # Searching for the last element
    mid = (lo+hi)//2
    if arr[mid] == value:
      last = mid
      lo = mid+1
    elif arr[mid] > value:
      hi = mid-1
    elif arr[mid] < value:
      lo = mid+1


  print("First occurrence:", first)
  print("Last occurrence:", last)

arr = np.array([1, 2, 2, 3, 3, 3, 3, 4, 5, 7, 3, 6, 6, 7])
first_last_occurence(arr, 3)
```

```python
# increasing (0->i-1) - increasing (i->N-1)
def find_transition_index(arr):
    left = 0
    right = len(arr) - 1
    while left < right:
        mid = (left + right) // 2
        if arr[mid] > arr[right]:
            left = mid + 1
        else:
            right = mid
    return left



arr = np.array([9, 12, 15, 17, 18, 1, 2, 4, 5, 7, 8])
i = find_transition_index(arr)
print("Transition index i =", i)
```

```python
# increasing (0->i-1) - decreasing (i->N-1)
def find_transition_index(arr):
    lo = 0
    hi = len(arr)-1

    while lo<=hi:
        mid = (lo+hi)//2
        if arr[mid-1] < arr[mid] < arr[mid+1]:
            lo = mid+1
        elif arr[mid-1] > arr[mid] > arr[mid+1]:
            hi = mid-1
        elif arr[mid]>arr[mid-1] and arr[mid]>arr[mid+1]:
            return mid



arr = [1, 3, 4, 5, 7, 9, 6, 2, -1, -3]
x = find_transition_index(arr)
print(x)
```

```python
# decreasing (0->i-1) - increasing (i->N-1)
def find_transition_index1(arr):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] < arr[mid - 1] and arr[mid] < arr[mid + 1]:
            return f"Transition found at index {mid}"
        elif arr[mid] > arr[mid + 1]:
            left = mid + 1
        elif arr[mid] < arr[mid+1]:
            right = mid - 1

    return "Not found"



arr = np.array([13, 11, 9, 7, 6, 4, 2, 1, 3, 5, 8])
i = find_transition_index1(arr)
print(i)
```
